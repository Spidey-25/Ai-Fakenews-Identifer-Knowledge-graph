```python
import pandas as pd
import numpy as np
from faker import Faker
from datetime import datetime, timedelta
import random

# Initialize faker
fake = Faker()

# Set random seed for reproducibility
random.seed(42)
np.random.seed(42)

# Create lists to store our data
titles = []
texts = []
sources = []
dates = []
labels = []  # 0 for real, 1 for fake
categories = []

# Common news categories
news_categories = ['Politics', 'Health', 'Technology', 'Business', 'Entertainment',
                   'Sports', 'Science', 'World', 'Environment', 'Education']

# Real news templates (these will be modified with realistic details)
real_templates = [
    "New study reveals {science_health_fact} according to researchers at {university}",
    "{company} announces {business_development} in quarterly earnings report",
    "{political_figure} proposes {policy} legislation to address {social_issue}",
    "Scientists discover {science_discovery} in {location}",
    "{sports_team} wins {sports_championship} after {dramatic_event}",
    "Global leaders meet to discuss {world_issue} at {conference_name}",
    "Breakthrough in {technology_field} could lead to {positive_outcome}",
    "{city} implements new program to tackle {local_problem}",
    "Annual report shows {economic_indicator} {trending_up_down} by {percentage}",
    "Experts warn about {environmental_concern} as new data shows {alarming_statistic}"
]

# Fake news templates
fake_templates = [
    "Shocking revelation: {conspiracy_target} secretly {outrageous_claim}",
    "You won't believe what {celebrity} said about {controversial_topic}",
    "{unlikely_cause} found to be the real reason behind {common_problem}",
    "Doctors hate this one trick for curing {disease} with {common_household_item}",
    "{political_figure} caught in scandal involving {shocking_behavior}",
    "Secret documents reveal {conspiracy_theory} covered up by {powerful_entity}",
    "{common_food_item} found to cause {serious_condition}, scientists say",
    "Ancient prophecy predicts {current_event} will lead to {apocalyptic_outcome}",
    "{celebrity_pair} secretly {implausible_relationship_status} for {time_period}",
    "Breaking: {mainstream_media} finally admits {fringe_belief} is true"
]

# Helper functions to generate realistic components
def generate_real_components():
    components = {
        'science_health_fact': random.choice([
            'benefits of moderate coffee consumption',
            'link between exercise and mental health',
            'potential treatment for Alzheimer\'s',
            'impact of sleep on productivity'
        ]),
        'university': fake.company_suffix() + ' University',
        'company': fake.company(),
        'business_development': random.choice([
            'record profits',
            'a new product line',
            'expansion into new markets',
            'a merger with ' + fake.company()
        ]),
        'political_figure': 'President ' + fake.last_name() if random.random() > 0.5 else 'Senator ' + fake.last_name(),
        'policy': random.choice([
            'new education',
            'healthcare reform',
            'climate change',
            'tax reform'
        ]),
        'social_issue': random.choice([
            'income inequality',
            'climate change',
            'healthcare access',
            'education reform'
        ]),
        'science_discovery': random.choice([
            'new species',
            'evidence of ancient civilization',
            'breakthrough in quantum computing',
            'promising cancer treatment'
        ]),
        'location': random.choice([
            'the Amazon rainforest',
            'the depths of the Pacific Ocean',
            'a remote region of Siberia',
            'an archaeological site in Egypt'
        ]),
        'sports_team': random.choice([
            'New York Titans',
            'London Royals',
            'Tokyo Samurai',
            'Berlin Bears'
        ]),
        'sports_championship': random.choice([
            'the championship',
            'the finals',
            'the World Series',
            'the Super Bowl'
        ]),
        'dramatic_event': random.choice([
            'a thrilling overtime victory',
            'a last-minute goal',
            'an unexpected comeback',
            'a record-breaking performance'
        ]),
        'world_issue': random.choice([
            'global warming',
            'refugee crises',
            'international trade',
            'nuclear disarmament'
        ]),
        'conference_name': random.choice([
            'Global Summit',
            'World Leaders Forum',
            'International Conference',
            'United Nations Assembly'
        ]),
        'technology_field': random.choice([
            'artificial intelligence',
            'renewable energy',
            'nanotechnology',
            'biomedical engineering'
        ]),
        'positive_outcome': random.choice([
            'revolutionary applications',
            'significant efficiency improvements',
            'breakthrough medical treatments',
            'solving critical environmental issues'
        ]),
        'city': fake.city(),
        'local_problem': random.choice([
            'homelessness',
            'traffic congestion',
            'affordable housing',
            'public transportation'
        ]),
        'economic_indicator': random.choice([
            'unemployment',
            'GDP growth',
            'inflation',
            'consumer confidence'
        ]),
        'trending_up_down': random.choice(['increased', 'decreased']),
        'percentage': f"{random.randint(1, 15)}%",
        'environmental_concern': random.choice([
            'rising sea levels',
            'deforestation',
            'plastic pollution',
            'species extinction'
        ]),
        'alarming_statistic': random.choice([
            'accelerated rates',
            'unprecedented levels',
            'worse than predicted',
            'critical thresholds being crossed'
        ])
    }
    return components

def generate_fake_components():
    components = {
        'conspiracy_target': random.choice([
            'the government',
            'big pharma',
            'the elite',
            'mainstream media',
            'celebrity'
        ]),
        'outrageous_claim': random.choice([
            'controls the weather',
            'is hiding alien technology',
            'knows the exact date of the apocalypse',
            'has been replaced by a clone'
        ]),
        'celebrity': fake.name(),
        'controversial_topic': random.choice([
            'vaccines',
            'the moon landing',
            'secret societies',
            'the flat earth'
        ]),
        'unlikely_cause': random.choice([
            '5G networks',
            'chemtrails',
            'secret government experiments',
            'ancient curses'
        ]),
        'common_problem': random.choice([
            'cancer rates',
            'mental illness',
            'climate change',
            'political polarization'
        ]),
        'disease': random.choice([
            'cancer',
            'diabetes',
            'COVID-19',
            'arthritis'
        ]),
        'common_household_item': random.choice([
            'baking soda',
            'vinegar',
            'garlic',
            'turmeric'
        ]),
        'political_figure': 'President ' + fake.last_name(),
        'shocking_behavior': random.choice([
            'secret offshore accounts',
            'illegal surveillance',
            'bribery scandal',
            'secret family'
        ]),
        'conspiracy_theory': random.choice([
            'the earth is flat',
            'aliens built the pyramids',
            'vaccines cause autism',
            'the moon landing was faked'
        ]),
        'powerful_entity': random.choice([
            'the Illuminati',
            'the deep state',
            'big tech',
            'global elites'
        ]),
        'common_food_item': random.choice([
            'avocados',
            'soy products',
            'gluten',
            'dairy'
        ]),
        'serious_condition': random.choice([
            'cancer',
            'autism',
            'infertility',
            'brain damage'
        ]),
        'current_event': random.choice([
            'the pandemic',
            'the election',
            'climate protests',
            'economic downturn'
        ]),
        'apocalyptic_outcome': random.choice([
            'the end of civilization',
            'a new world order',
            'mass extinction',
            'total societal collapse'
        ]),
        'celebrity_pair': f"{fake.name()} and {fake.name()}",
        'implausible_relationship_status': random.choice([
            'married',
            'feuding',
            'working together secretly',
            'related'
        ]),
        'time_period': random.choice([
            'years',
            'decades',
            'centuries',
            'since childhood'
        ]),
        'mainstream_media': random.choice([
            'CNN',
            'Fox News',
            'BBC',
            'The New York Times'
        ]),
        'fringe_belief': random.choice([
            'chemtrails are real',
            'the earth is hollow',
            'time travel exists',
            'shape-shifting reptilians control the government'
        ])
    }
    return components

# Generate 1000 real news samples
for _ in range(1000):
    template = random.choice(real_templates)
    components = generate_real_components()
    title = template.format(**components)
    titles.append(title)

    # Generate longer text content
    text = f"In a {random.choice(['groundbreaking', 'significant', 'important'])} development, {title.lower()}. "
    text += fake.paragraph(nb_sentences=5) + " "
    text += fake.paragraph(nb_sentences=4)
    texts.append(text)

    sources.append(random.choice([
        'Reuters', 'Associated Press', 'BBC News', 'The New York Times',
        'The Washington Post', 'The Guardian', 'NPR', 'PBS NewsHour'
    ]))

    dates.append(fake.date_between(start_date='-2y', end_date='today').strftime('%Y-%m-%d'))
    labels.append(0)  # Real news
    categories.append(random.choice(news_categories))

# Generate 1000 fake news samples
for _ in range(1000):
    template = random.choice(fake_templates)
    components = generate_fake_components()
    title = template.format(**components)
    titles.append(title)

    # Generate longer text content
    text = f"EXCLUSIVE: {title.upper()}. {fake.last_name()}, a {random.choice(['renowned', 'controversial', 'secret'])} "
    text += f"{random.choice(['scientist', 'whistleblower', 'insider', 'expert'])} reveals shocking details. "
    text += fake.paragraph(nb_sentences=4) + " "
    text += fake.paragraph(nb_sentences=3) + " "
    text += "CLICK HERE TO LEARN MORE AND SHARE THIS URGENT NEWS!"
    texts.append(text)

    sources.append(random.choice([
        'TruthSeeker Daily', 'Real News Network', 'The Patriot Report',
        'Freedom Times', 'The People\'s Voice', 'Wake Up America News',
        'The Liberty Beacon', 'Natural News', 'InfoWars', 'Before It\'s News'
    ]))

    dates.append(fake.date_between(start_date='-2y', end_date='today').strftime('%Y-%m-%d'))
    labels.append(1)  # Fake news
    categories.append(random.choice(news_categories))

# Shuffle the dataset to mix real and fake news
combined = list(zip(titles, texts, sources, dates, labels, categories))
random.shuffle(combined)
titiles, texts, sources, dates, labels, categories = zip(*combined)

# Create DataFrame
df = pd.DataFrame({
    'title': titiles,
    'text': texts,
    'source': sources,
    'date': dates,
    'label': labels,
    'category': categories
})

# Add description
df.attrs['description'] = 'Smoke & Mirrors News Collection - A synthetic dataset for fake news detection research'

# Save to CSV
df.to_csv('Smoke & Mirrors News Collection.csv', index=False)

print("Dataset successfully created and saved as 'Smoke & Mirrors News Collection.csv'")
print(f"Total entries: {len(df)} (Real: {len([x for x in labels if x == 0])}, Fake: {len([x for x in labels if x == 1])})")
```
